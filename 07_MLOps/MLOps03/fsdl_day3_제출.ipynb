**✅ Chapter 13 체크포인트**
1. ML 테스트가 "다크 소울"이라 불리는 이유를 전통 소프트웨어와 비교하여 3가지 관점에서 설명할 수 있으신가요?
   - **ML 테스트가 '다크 소울'이라 불리는 이유:**
     - **불확실성:** ML 모델은 본질적으로 확률적이며, 100% 정확한 경우가 드물어 테스트 결과의 해석이 복잡합니다.
     - **변동성:** 데이터와 모델 학습 과정의 작은 변화에도 성능이 크게 달라질 수 있어 일관된 테스트 결과 확보가 어렵습니다.
     - **난해한 디버깅:** 실패 원인이 모델 아키텍처, 데이터, 하이퍼파라미터 등 다양하며, 직관적인 디버깅이 어렵습니다.
2. 테스트 수트를 "분류기"로 보았을 때, 오탐(False Alarm)이 왜 위험한지 실제 사례를 들어 설명할 수 있으신가요?
   - **테스트 수트를 '분류기'로 보았을 때 오탐(False Alarm)이 위험한 이유:**
     - 오탐은 실제로는 문제가 없는 좋은 모델 변경을 '실패'로 분류하여 개발자가 불필요한 디버깅에 시간을 낭비하게 만듭니다. 이는 개발 속도를 저해하고, 결국 개발자들이 테스트를 신뢰하지 않게 되어 무시하는 결과를 초래할 수 있습니다. 중요한 버그가 묻히거나, 유용한 개선이 배포되지 못하게 됩니다.
3. 암기 테스트(`overfit_batches=1`)가 실패했을 때 의심해야 할 원인 5가지를 나열하고, 각각의 디버깅 방법을 떠올릴 수 있으신가요?
   - **암기 테스트(`overfit_batches=1`) 실패 시 의심할 원인 5가지:**
     - **학습률이 너무 높거나 낮음:** 학습률을 조정하여 최적의 값을 찾습니다.
     - **모델 복잡도가 충분하지 않음:** 모델 레이어를 추가하거나 노드 수를 늘려 표현력을 높입니다.
     - **손실 함수 또는 옵티마이저 문제:** 손실 함수 정의와 옵티마이저 설정을 검토합니다.
     - **데이터 문제:** 배치 크기, 데이터 정규화, 데이터 로딩 파이프라인을 확인합니다.
     - **GPU/TPU 설정 오류:** 하드웨어 가속기가 올바르게 작동하는지 확인합니다.
4. 데이터 기댓값 테스트에서 "느슨한 경계" 원칙이 왜 중요한지, 경계를 너무 엄격하게 잡으면 어떤 문제가 발생하는지 설명해 보세요.
   - **데이터 기댓값 테스트에서 '느슨한 경계' 원칙이 중요한 이유:**
     - ML 시스템의 데이터는 본질적으로 변동성이 크고 예상치 못한 패턴을 포함할 수 있기 때문에 엄격한 경계는 잦은 오탐을 유발합니다. 느슨한 경계는 불필요한 경고를 줄여 테스트의 신뢰도를 높이고, 개발자가 정말 중요한 데이터 이상에 집중할 수 있도록 돕습니다. 너무 엄격한 경계는 개발자들이 테스트를 무시하게 만들 수 있습니다.
5. 행동 테스트의 "불변성 테스트"와 "방향성 테스트"의 차이를 자신의 프로젝트에 적용하여 각각 1개씩 예시를 만들어 보실 수 있나요?
   - **행동 테스트 예시:**
     - **불변성 테스트:** 챗봇의 긍정적인 답변(`'좋아요!'`)에 오탈자를 추가(`'조아요!'`)하더라도, 여전히 긍정적인 응답을 유지하는지 확인합니다.
     - **방향성 테스트:** 가격 예측 모델에서 `방 크기`를 증가시켰을 때, `예측 가격`이 증가하는(혹은 최소한 감소하지 않는) 방향성을 유지하는지 확인합니다.

**✅ Chapter 14 체크포인트**
1. "Make it Run → Make it Fast → Make it Right" 순서에서, 왜 ML에서는 "Fast"가 "Right"보다 먼저 와야 하는지 설명할 수 있으신가요?
   - **답변:** ML에서는 'Fast'가 'Right'보다 먼저 와야 하는 이유는 빠른 반복과 실험의 중요성 때문입니다. 일단 작동하는(Run) 모델을 빠르게 만들고(Fast) 이를 통해 가설을 검증하며 피드백을 받아야 합니다. 처음부터 완벽한(Right) 모델을 만들려고 하면 개발 주기가 매우 길어지고 비용이 많이 들어 실제 작동 가능한 모델을 찾기 어렵기 때문에, '빠르게 만들고 개선하는' 전략이 훨씬 효과적입니다.
2. OOM 오류를 해결하는 4단계 전략을 순서대로 말할 수 있으신가요? Adam이 파라미터의 3~4배 메모리를 쓰는 이유는 무엇인가요?
   - **답변:** OOM(Out of Memory) 오류 해결 4단계 전략은 일반적으로 다음과 같습니다: 1. 배치 크기 줄이기 2. 모델 크기(파라미터 수) 줄이기 3. 정밀도 낮추기(예: `float32`를 `float16`으로) 4. CPU 오프로딩이나 그라디언트 체크포인팅과 같은 고급 기법 사용하기.  
   Adam 옵티마이저가 파라미터의 3~4배 메모리를 사용하는 이유는 각 파라미터에 대해 '과거 그래디언트의 이동 평균'과 '과거 그래디언트 제곱의 이동 평균'이라는 두 가지 추가 상태 변수를 저장하기 때문입니다. 이러한 모멘텀 상태들이 파라미터와 함께 메모리를 차지하여 총 메모리 사용량이 증가합니다.
3. 프로파일링 결과에서 `get_train_batch`가 전체 시간의 60%를 차지한다면, 어떤 최적화를 우선 시도하시겠어요? 반대로 `training_step`이 90%라면요?
   - **답변:** 프로파일링 결과 `get_train_batch`가 전체 시간의 60%를 차지한다면, 데이터 로딩 및 전처리 과정이 병목이므로 `DataLoader`의 `num_workers`를 늘리거나 데이터 전처리 속도를 최적화하는 방안을 우선 시도할 것입니다. 반대로 `training_step`이 90%를 차지한다면, 모델의 순전파 및 역전파 계산이 병목이므로 모델 복잡도를 줄이거나, 혼합 정밀도 학습을 적용하거나, GPU 활용을 최적화하는 방안을 고려할 것입니다.
4. 스케일링 법칙의 "작은 실험 → 직선 외삽 → 필요 자원 추산" 흐름을 자신의 언어로 설명하고, 이것이 비즈니스적으로 어떤 가치를 주는지 설명해 보세요.
   - **답변:** 스케일링 법칙의 흐름은 먼저 작은 규모의 실험 여러 개를 수행하여 컴퓨팅 자원, 데이터셋 크기, 모델 파라미터 수와 모델 성능(손실) 사이의 관계를 파악하는 것입니다. 이 결과를 로그-로그 스케일 그래프에 표시하면 직선 형태의 관계를 발견할 수 있고, 이 직선을 연장하여 목표 성능에 도달하기 위해 필요한 자원을 수학적으로 추정할 수 있습니다. 비즈니스적으로는 자원 낭비를 줄이고 투자 대비 효과를 극대화하는 데 큰 가치를 제공합니다. 막연한 추측 대신 데이터 기반으로 최적의 투자 지점을 파악하고, 예측 가능한 성능 개선 로드맵을 수립할 수 있게 해줍니다.
5. 성능 저하 진단표(오버피팅/언더피팅/분포 변화)에서, 자신의 최근 프로젝트가 어떤 상황이었는지 분석하고 적절한 처방을 매칭해 보세요.
   - **답변:**
     - **오버피팅(Overfitting):** 훈련 데이터에서는 성능이 매우 좋지만 검증/테스트 데이터에서는 저조할 때 발생합니다. 처방으로는 더 많은 데이터 확보, 정규화(L1, L2), 드롭아웃 적용, 조기 종료, 모델 복잡도 줄이기 등이 있습니다.
     - **언더피팅(Underfitting):** 훈련 데이터와 검증/테스트 데이터 모두에서 성능이 낮을 때 발생합니다. 처방으로는 모델 복잡도 높이기, 학습 시간 늘리기, 더 나은 특징 엔지니어링, 정규화 강도 줄이기 등이 있습니다.
     - **데이터 분포 변화(Data Distribution Shift):** 학습 데이터와 실제 서비스 데이터의 분포가 달라져 모델 성능이 저하될 때 발생합니다. 처방으로는 최신 데이터로 모델 재학습, 도메인 적응(Domain Adaptation) 기법 적용, 또는 분포 변화에 강건한 특징을 사용하는 방법 등이 있습니다.

1. LLM 기반 시스템을 Accuracy만으로 평가할 수 없는 이유를 설명하고, 추가로 필요한 평가 차원을 5가지 나열할 수 있으신가요?
   - **답변:** LLM 기반 시스템은 단순히 정답 여부를 넘어 답변의 품질, 안전성 등 다양한 측면이 중요하기 때문에 Accuracy만으로는 평가할 수 없습니다. 추가로 필요한 평가 차원으로는 유창성(Fluency), 사실성(Faithfulness), 스타일(Tone), 의도 이해(Intent Understanding), 안전성(Safety) 등이 있습니다.
2. LLM-as-a-Judge의 장점과 주의할 한계(편향)를 각각 설명하고, 편향을 완화하는 방법을 하나 제안해 보세요.
   - **답변:** LLM-as-a-Judge의 장점은 평가 프로세스를 자동화하여 빠르고 효율적인 피드백 루프를 만들 수 있다는 것입니다. 하지만 LLM 평가자 자체의 편향(예: 특정 답변 스타일 선호, 특정 LLM에 대한 선호)이 존재할 수 있다는 한계가 있습니다. 이러한 편향을 완화하기 위해 평가 프롬프트에 다양한 관점을 제시하거나, 여러 LLM 평가자의 결과를 앙상블하는 방법을 사용할 수 있습니다.
3. 입력 드리프트, 출력 드리프트, 성능 드리프트의 차이를 각각 한 문장으로 설명하고, 자신의 프로젝트에서 가장 가능성 높은 드리프트 유형은 무엇인지 분석해 보세요.
   - **답변:** **입력 드리프트**는 모델에 들어오는 데이터 분포가 학습 시점과 달라지는 현상입니다. **출력 드리프트**는 모델의 예측 분포가 변하는 것입니다. **성능 드리프트**는 실제 정답을 기준으로 모델의 성능이 저하되는 현상입니다. (이 부분은 '자신의 프로젝트'에 대한 답변은 AI가 할 수 없으므로 일반적인 답변으로 대체합니다) 제 가상의 프로젝트라면, 사용자 질의의 트렌드 변화로 인한 '입력 드리프트'가 가장 가능성 높을 것입니다.
4. "모델은 시간이 지나면 썩는다"는 말의 의미를 코로나19 같은 구체적 사례로 설명할 수 있으신가요? 이를 방지하기 위해 어떤 시스템이 필요한가요?
   - **답변:** "모델은 시간이 지나면 썩는다"는 것은 학습 당시의 데이터 패턴이 실제 환경과 달라지면서 모델의 예측 성능이 저하된다는 의미입니다. 코로나19 팬데믹으로 인해 소비 패턴, 교통량 등 사회 전반의 데이터 분포가 급변하여, 이전 데이터로 학습된 모델들의 성능이 크게 저하된 것이 대표적인 사례입니다. 이를 방지하기 위해서는 지속적인 **모니터링 시스템**을 통해 데이터 드리프트 및 성능 저하를 감지하고, 필요시 모델을 재학습하거나 업데이트하는 **자동화된 재학습 파이프라인**이 필요합니다.
