## [Q4] 파이썬을 사용하여 간단한 계산기 프로그램 만들기

```python
# math 모듈 import
import math

# 계산기 함수 생성
def calculator():
    while True:
        # 첫 번째 정수 입력 (ValueError 처리 및 반복)
        while True:
            try:
                num1 = int(input("첫 번째 정수를 입력하세요: "))
                break  # 정수 입력 시 다음 단계로
            except ValueError:
                print("잘못된 입력입니다. 정수를 입력해주세요.")
                
        # 두 번째 정수 입력받기 (ValueError 처리 및 반복)
        while True:
            try:
                num2 = int(input("두 번째 정수를 입력하세요: "))
                break  # 정수 입력 시 다음 단계로
            except ValueError:
                print("잘못된 입력입니다. 정수를 입력해주세요.")
        
        #연산자 입력받기
        operator = input("연산자를 입력하세요 (+, -, *, /, **): ").strip()
        
        #계산 및 예외 처리
        try:
            if operator == '+':
                result = num1 + num2
                print(f"결과: {num1} + {num2} = {result}")
                
            elif operator == '-':
                result = num1 - num2
                print(f"결과: {num1} - {num2} = {result}")
                
            elif operator == '*':
                result = num1 * num2
                print(f"결과: {num1} * {num2} = {result}")
                
            elif operator == '/':
                result = num1 / num2
                print(f"결과: {num1} / {num2} = {result}")
                
            elif operator == '**':
                # math 모듈을 사용한 제곱 연산
                result = math.pow(num1, num2)
                print(f"결과: {num1} ** {num2} = {result}")
                
            else:
                # 지원하지 않는 연산자 처리
                print("지원하지 않는 연산자입니다. (+, -, *, /, ** 중 하나를 입력해주세요)")

        # 두 번째 수가 0일 때 오류 처리
        except ZeroDivisionError:
            print("오류: 0으로 나눌 수 없습니다.")

        # 구분선
        print("-" * 40)
        
        # 계속 진행 여부 확인
        continue_calc = input("계속 계산을 하시겠습니까? (y/n): ")
        if continue_calc == 'n':
            print("계산기를 종료합니다. 감사합니다!")
            break
        elif continue_calc == 'y':
            print("\n새로운 계산을 시작합니다.")
calculator()
```
