## [Q1] 평행사변형, 사다리꼴 넓이 구하기

```python
# 사각형 넓이를 구하는 클래스 완성!
class Square:
    def __init__(self):
        self.square = int(input('넓이를 구하고 싶은 사각형의 숫자를 써주세요.\n 1.직사각형 2.평행사변형 3.사다리꼴 \n >>>'))

        if self.square == 1:
            print('직사각형 함수는 rect()입니다.')

        elif self.square == 2:
            print('평행사변형 함수는 par()입니다.')
        
        elif self.square == 3:
            print('사다리꼴 함수는 trape()입니다.')
        
        else:
            print('1, 2, 3 중에서 다시 입력해주세요')

    def rect(self):
        width, vertical = map(int, input('가로, 세로를 입력하세요. 예시 : 가로,세로\n >>>').split(','))
        area = width * vertical
        result = '직사각형의 넓이는 : ' + str(area)
        return result

    def par(self):
        bottom, height = map(int, input('밑변, 높이를 입력하세요. 예시 : 밑변,높이\n >>>').split(','))
        area = bottom * height
        result = '평행사변형의 넓이는 : ' + str(area)
        return result

    def trape(self):
        bottom, top, height = map(int, input('밑변, 윗변, 높이를 입력하세요. 예시 : 밑변,윗변,높이\n >>>').split(','))
        area = (bottom + top) * height / 2
        result = '사다리꼴의 넓이는 : ' + str(area)
        return result

a = Square()  # 객체 생성 & 2, 3을 각각 입력해 봅시다.
a.rect()
a.par()
a.trape()
```

## [Q2] 키오스크(Kiosk) 만들기

```python
class Kiosk:
    def __init__(self):
        self.menu =  ['americano', 'latte', 'mocha', 'yuza_tea', 'green_tea', 'choco_latte']
        self.price = [2000, 3000, 3000, 2500, 2500, 3000]
        # 주문을 담을 빈 리스트 생성
        self.order_menu = [] 
        self.order_price = []

    # 메뉴 출력 메서드
    def menu_print(self):
        for i in range(len(self.menu)):
            print(i+1, self.menu[i], ':', self.price[i], '원')

    # 주문 메서드
    def menu_select(self):
        print() #줄바꿈
        n = 0
        while n < 1 or len(self.menu) < n:
            n = int(input("음료 번호를 입력하세요 : ")) 

            #메뉴판에 있는 번호일 때 
            if 1 <= n and n <= len(self.menu):
                self.n = n 
                self.price_sum = self.price[n-1] 

            #메뉴판에 없는 번호일 때
            else:
                print("없는 메뉴입니다. 다시 주문해 주세요.")

        # 음료 온도 물어보기
        t = 0  
        while t != 1 and t != 2: 
            t= int(input("HOT 음료는 1을, ICE 음료는 2를 입력하세요 : "))
            if t == 1:
                self.temp = "HOT"
            elif t == 2:
                self.temp = "ICE"
            else:    
                print("1과 2 중 하나를 입력하세요.\n")
        
        # 주문 내용을 출력 
        print('주문 음료', self.temp, self.menu[n-1], ' : ', self.price[n-1], '원')
        
        # 첫 번째 주문을 장바구니에 담기
        self.order_price.append(self.price[n-1])
        self.order_menu.append(self.temp + ' ' + self.menu[n-1])
        
        #추가 주문 또는 지불
        while n != 0: #지불 선택 전까지 반복
            print() 
            n = int(input("추가 주문은 음료 번호를, 지불은 0을 누르세요 : ")) 
            if n > 0  and n < len(self.menu) +1:
                #추가 음료 온도
                t = 0
                while t != 1 and t != 2:
                    t= int(input("HOT 음료는 1을, ICE 음료는 2를 입력하세요 : "))
                    if t == 1:
                        self.temp = "HOT"
                    elif t == 2:
                        self.temp = "ICE"
                    else:
                        print("1과 2 중 하나를 입력하세요.\n")                
                
                # 주문 리스트 추가
                self.order_price.append(self.price[n-1]) #가격 리스트에 추가
                self.order_menu.append(self.temp + ' ' + self.menu[n-1]) #주문 리스트에 추가
                self.price_sum += self.price[n-1] #합계 금액

                # 추가주문 내용을 출력 
                print('추가주문 음료', self.temp, self.menu[n-1], ' : ', self.price[n-1], '원') 
                
        else : 
            if n == 0 :
                print("주문이 완료되었습니다.")
                print(self.order_menu, self.order_price)
            else : 
                print("없는 메뉴입니다.")
            
    # 지불 메서드
    def pay(self):
        #합계 금액 출력
        print(f"\n총 합계 금액은 {self.price_sum}원 입니다.")
        while True:
            pay_type = input("결제 수단을 선택해주세요 (1 또는 cash: 현금, 2 또는 card: 카드) : ").strip().lower()
            
            # 현금 결제인 경우
            if pay_type == '1' or pay_type == 'cash':
                print('직원을 호출하겠습니다.')
                break  
                
            # 카드 결제인 경우
            elif pay_type == '2' or pay_type == 'card':
                print('IC칩 방향에 맞게 카드를 꽂아주세요.')
                break  
                
            # 그 외의 잘못된 값을 입력했을 경우
            else:
                print('잘못된 입력입니다. 다시 선택해주세요.\n')

    # 주문서 출력 
    def table(self):
        # 외곽
        print('\n⟝' + '-' * 30 + '⟞')
        for i in range(2):
            print('|' + ' ' * 30 + '|')

        # 주문 상품명 : 해당 금액
        for i in range(len(self.order_menu)):
            print(f"| {self.order_menu[i]} : {self.order_price[i]}원")
        
        print(f"| \n| 총 주문 금액 : {self.price_sum}원")

        # 외곽
        for i in range(2):
            print('|' + ' ' * 30 + '|')
        print('⟝' + '-' * 30 + '⟞')

a = Kiosk()  # 객체 생성 
a.menu_print()  # 메뉴 출력
a.menu_select()  # 주문
a.pay()  # 지불
a.table()  # 주문표 출력
```
