## [Q1-1]회문 판독기

```python
def is_palindrome(target):
    return target == target[::-1]

target1 = "우영우우영우"
target2 = "기러기기러기"

print(is_palindrome(target1))
print(is_palindrome(target2))
```

## [Q1-2]회문 분류기

```python
def is_palindrome(target):
    return target == target[::-1]
    
def palindrome_classifier(target_list):
    result_dict = {
        "회문O": [],
        "회문X": []
    }

    for target in target_list:
            if is_palindrome(target):
                result_dict["회문O"].append(target)
            else:
                result_dict["회문X"].append(target)

    total_count = len(target_list)
    palindrome_count = len(result_dict["회문O"])

    palindrome_ratio = palindrome_count / total_count if total_count > 0 else 0.0

    return result_dict, palindrome_count, palindrome_ratio


targets = ['기러기', '기러귀', '토마토', '토마톱', '스위스', '스위수', '우영우', '우영웃']
print(palindrome_classifier(targets))
```
