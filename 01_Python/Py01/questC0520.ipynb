## [Q1-1]회문 판독기

```python
def is_palindrome(target):
    return target == target[::-1]

target1 = "아많다많다많다많아"
target2 = "아정말많다많다많아"

print(is_palindrome(target1))
print(is_palindrome(target2))
```

## [Q1-2]회문 분류기

```python
def is_palindrome(target):
    return target == target[::-1]
    
def palindrome_classifier(target_list):
    result_dict = {
        "회문O": [],
        "회문X": []
    }

    for target in target_list:
            if is_palindrome(target):
                result_dict["회문O"].append(target)
            else:
                result_dict["회문X"].append(target)

    total_count = len(target_list)
    palindrome_count = len(result_dict["회문O"])

    palindrome_ratio = palindrome_count / total_count if total_count > 0 else 0.0

    return result_dict, palindrome_count, palindrome_ratio


targets = ['기러기', '까마귀', '토마토', '스윗소스', '스위스', '인도인', '한국인', '별똥별', '우영우', '우향우']
print(palindrome_classifier(targets))
```

## [Q2-2]성적 분석기 (미완료)

```python
scores = {
    "유원영": {"국어": 94, "영어": 91, "수학": 89},
    "설지민": {"국어": 88, "영어": 96, "수학": 90},
    "신윤아": {"국어": 86, "영어": 88, "수학": 77},
    "장윤아": {"국어": 98, "영어": 76, "수학": 92}
}
def score_analysis(name):
    subjects = ["국어", "영어", "수학"]
    score = scores[name]
    class_avg = {}
    
    # 1. 과목별 학급 평균 계산 (리스트 컴프리헨션 활용)
    for x in subjects:
        class_avg[x] = sum([student_score[x] for student_score in scores.values()]) / len(scores)
        
    # 2. [양식 조립 시작] 화면에 예쁘게 출력하기
    print(f"[{name} 성적 분석]")
    
    student_total = 0
    class_total = 0
    
    for x in subjects:
        st_score = score[x]        # 학생 점수
        bg_avg = class_avg[x]      # 학급 평균
        diff = st_score - bg_avg   # 대비 점수 (+/- 판별용)
        
        # 대비 점수 부호(+ 또는 없으면 생략) 결정 규칙
        sign = "+" if diff >= 0 else ""
        
        print(f"- {x} : {st_score}점 / 학급평균({bg_avg:.2f}점) 대비 {sign}{diff:.2f}점")
        
        # 전체 과목 합산을 위해 더하기 상자에 누적
        student_total += st_score
        class_total += bg_avg
        
    # 3. 전과목 평균 계산 및 출력
    student_avg = student_total / len(subjects)
    class_total_avg = class_total / len(subjects)
    total_diff = student_avg - class_total_avg
    total_sign = "+" if total_diff >= 0 else ""
    
    print(f"- 전과목 평균 : {student_avg:.2f}점 / 학급평균({class_total_avg:.2f}점) 대비 {total_sign}{total_diff:.2f}점")

    
score_analysis("설지민")
```
