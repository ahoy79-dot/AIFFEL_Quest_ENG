### ✅ 체크포인트 01


1. FastAPI가 Flask보다 모델 배포에 적합한 이유 세 가지는 무엇입니까?
    * **답변**: 높은 성능 (Starlette 기반), 자동 데이터 유효성 검사 및 직렬화 (Pydantic), 그리고 자동 API 문서 생성 (Swagger UI/ReDoc) 덕분에 모델 배포에 적합합니다.
2. Uvicorn의 역할은 무엇이며, 왜 FastAPI와 함께 사용합니까?
    * **답변**: Uvicorn은 ASGI(Asynchronous Server Gateway Interface) 서버로, FastAPI 애플리케이션을 실행하는 고성능 웹 서버입니다. FastAPI는 프레임워크이므로, HTTP 요청 및 응답을 처리하기 위해 Uvicorn과 같은 ASGI 서버가 필요합니다.
3. `@app.get("/health")`에서 `get`과 `"/health"`는 각각 무엇을 의미합니까?
    * **답변**:
        * `get`: 해당 엔드포인트가 HTTP GET 요청을 처리함을 나타냅니다.
        * `"/health"`: 이 엔드포인트의 URL 경로를 `/health`로 정의합니다.
4. FastAPI에서 dict를 반환하면 어떤 일이 자동으로 일어납니까?
    * **답변**: FastAPI는 파이썬 딕셔너리를 자동으로 JSON 응답으로 변환합니다.

### ✅ 체크포인트 02

1. `/models/sentiment-v1`에서 `sentiment-v1`은 어떤 종류의 파라미터입니까?
    * **답변**: Path 파라미터입니다.
2. `/models?status=running&limit=5`에서 `status`와 `limit`은 어떤 종류의 파라미터입니까?
    * **답변**: Query 파라미터입니다.
3. 모델 추론 요청에 Request Body를 사용하는 이유는 무엇입니까?
    * **답변**: 입력 데이터가 복잡하고 구조화되어 있거나, 여러 필드를 전달해야 하거나, 데이터 양이 많을 때 (URL 길이 제한 없음), 또는 중첩된 데이터 구조를 보낼 때 적합하기 때문입니다.
4. FastAPI에서 함수의 파라미터가 Path, Query, Body 중 어디서 오는지 어떻게 판별합니까?
    * **답변**: FastAPI는 함수의 파라미터가 타입 힌트와 기본값의 유무에 따라 자동으로 Path, Query, Body 파라미터를 판별합니다. 예를 들어, 경로에 포함된 파라미터는 Path, 기본값이 있는 파라미터는 Query, Pydantic 모델 타입으로 선언된 파라미터는 Request Body로 간주합니다.


### ✅ 체크포인트 03

1. FastAPI에서 Swagger UI에 접속하려면 어떤 URL로 이동합니까?
    * **답변**: `/docs` URL로 이동합니다.
2. Swagger UI가 코드와 항상 동기화될 수 있는 이유는 무엇입까?
    * **답변**: FastAPI는 개발자가 작성한 파이썬 코드를 기반으로 OpenAPI(Swagger) 사양을 자동으로 생성합니다. 따라서 코드를 변경하면 OpenAPI 사양과 Swagger UI도 자동으로 업데이트되어 항상 동기화됩니다.
3. Pydantic 모델의 `Field(description=, examples=)`는 Swagger UI의 어디에 반영됩니까?
    * **답변**: `description`은 해당 필드의 설명으로, `examples`는 해당 필드의 예시 값으로 Swagger UI에 반영됩니다.
4. Swagger UI와 ReDoc의 핵심 차이는 무엇입니까?
    * **답변**: Swagger UI는 API를 시각적으로 탐색하고 **직접 호출하여 테스트**할 수 있는 대화형 문서입니다. 반면 ReDoc은 더 **읽기 쉬운 형태**로 API 문서를 제공하는 데 중점을 둔 정적인 문서입니다. (테스트 기능 없음)


### ✅ 체크포인트 04

1. `text: str`과 `text: str = "기본값"`의 차이는 무엇입니까?
    * **답변**: `text: str`은 해당 필드가 **필수(required)** 임을 의미하며, 클라이언트가 반드시 값을 제공해야 합니다. `text: str = "기본값"`은 해당 필드가 **선택 사항(optional)** 이며, 클라이언트가 값을 제공하지 않으면 "기본값"이 사용됩니다.
2. `Field(..., min_length=1, max_length=5000)`에서 `...`은 무엇을 의미합니까?
    * **답변**: `...` (Ellipsis)는 Pydantic에서 해당 필드가 **필수(required)** 임을 명시적으로 나타내는 데 사용됩니다. 이 필드는 기본값이 없지만 반드시 제공되어야 함을 의미합니다.
3. 422 에러 응답에서 `loc` 필드는 어떤 정보를 담고 있습니까?
    * **답변**: `loc` 필드는 에러가 발생한 위치를 나타냅니다. 예를 들어 `["body", "text"]`는 요청 본문(body)의 `text` 필드에서 에러가 발생했음을 의미합니다.
4. `response_model`을 지정하면 어떤 이점이 있습니까?
    * **답변**:
        * **데이터 직렬화**: 반환되는 데이터가 `response_model`에 정의된 Pydantic 모델에 따라 자동으로 직렬화(JSON 변환)됩니다.
        * **데이터 유효성 검사**: 반환 데이터가 `response_model` 스키마를 따르는지 검사하여 잘못된 형식의 데이터가 클라이언트에 전달되는 것을 방지합니다.
        * **자동 문서화**: Swagger UI에 API 응답 스키마가 자동으로 반영되어 클라이언트 개발자가 응답 구조를 쉽게 이해할 수 있습니다.

### ✅ 체크포인트  05  

1. 모델을 서버 시작 시 한 번만 로드해야 하는 이유는 무엇입니까?  
    * **답변**: 모델을 서버 시작 시 한 번만 로드해야 하는 주된 이유는 성능 최적화와 리소스 효율성 때문입니다. 각 요청마다 모델을 로드하면 불필요한 지연과 높은 리소스 사용이 발생하여 서버 처리량을 저하시킵니다.
2. pixel_values가 784개가 아닌 요청이 들어오면 어떤 일이 발생합니까? 이를 처리하는 코드를 직접 작성했습니까?  
    * **답변**: `pixel_values`가 784개가 아닌 요청이 들어오면 Pydantic 모델의 `MinLen`과 `MaxLen` 유효성 검사에 의해 422 Unprocessable Entity 에러가 발생합니다. 이 에러 처리는 Pydantic 모델에 정의된 유효성 검사 규칙에 따라 FastAPI가 자동으로 처리하므로, 별도의 에러 처리 코드를 직접 작성하지 않아도 됩니다.
3. HTTPException(status_code=503)은 어떤 상황에서 사용했습니까? 왜 500이 아니라 503입니까?  
    * **답변**: `HTTPException(status_code=503)`은 모델 로딩에 실패하여 추론을 할 수 없는 상황에서 사용했습니다. 503은 "Service Unavailable"을 의미하며, 일시적으로 서비스를 제공할 수 없을 때 사용합니다. 500은 "Internal Server Error"로 서버 내부에서 예상치 못한 에러가 발생했을 때 사용하는데, 모델 로딩 실패는 서버가 의도적으로 서비스를 중단한 상황이므로 503이 더 적절합니다.
4. Swagger UI에서 PredictRequest의 description과 examples가 어디에 표시됩니까?
    * **답변**: Swagger UI에서 PredictRequest의 description과 examples는 API 요청 본문(Request Body) 스키마의 해당 필드 옆에 설명과 예시 값으로 각각 표시됩니다. 'Try it out' 섹션에서 요청을 보낼 때 예시 값을 확인할 수 있습니다.
