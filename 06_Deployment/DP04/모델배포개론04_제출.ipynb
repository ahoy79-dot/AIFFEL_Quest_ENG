### ✅ 체크포인트 01

1. Streamlit의 스크립트 재실행 모델이란 무엇입니까?
   - **답변**: 사용자의 어떤 상호작용(버튼 클릭, 입력 변경 등)이 발생할 때마다 Python 스크립트 전체를 위에서 아래로 처음부터 다시 실행하는 Streamlit의 동작 방식입니다.
2. `st.text_input()`에 값을 입력하면 내부적으로 어떤 일이 일어납니까?
   - **답변**: 사용자가 `st.text_input()` 위젯에 값을 입력하면, Streamlit 앱의 스크립트가 처음부터 다시 실행되고, `st.text_input()`은 새롭게 입력된 값을 반환합니다.
3. `st.set_page_config()`를 스크립트 중간에 호출하면 어떻게 됩니까?
   - **답변**: `st.set_page_config()`는 **반드시 스크립트의 가장 첫 번째 Streamlit 호출**이어야 합니다. 스크립트 중간에 호출하면 오류가 발생하거나 원하는 대로 동작하지 않을 수 있습니다.


### ✅ 체크포인트 02

1. `st.file_uploader()`로 업로드된 파일의 바이트 데이터는 어떻게 얻습니까?
   - **답변**: `uploaded_file.getvalue()` 메서드를 사용하여 파일의 바이트 데이터를 얻을 수 있습니다.
2. Streamlit에서 `@st.cache_resource`를 사용하는 이유는 무엇입니까?
   - **답변**: 스크립트가 재실행될 때마다 비용이 많이 드는 객체(예: 모델, API 클라이언트)를 다시 생성하지 않고, 한 번만 생성하여 메모리에 캐시하고 재사용하기 위함입니다. 이는 성능 최적화에 도움이 됩니다.


### ✅ 체크포인트 03

1. 모놀리식과 분리 아키텍처의 핵심 차이를 한 문장으로 설명하세요.
   - **답변**: 모놀리식은 모든 기능이 하나의 덩어리로 묶여 있는 반면, 분리 아키텍처는 프론트엔드와 백엔드가 독립적인 서비스로 분리되어 있습니다.
2. 모델을 업데이트할 때, 분리 아키텍처에서는 어떤 서버만 재배포하면 됩니까?
   - **답변**: 모델을 업데이트할 때는 FastAPI 백엔드 서버만 재배포하면 됩니다.
3. Streamlit 앱에 PyTorch가 설치되어 있지 않아도 되는 이유는 무엇입니까?
   - **답변**: Streamlit 앱은 FastAPI 백엔드 서버에 HTTP 요청을 보내고 JSON 응답을 받을 뿐, 직접 모델 추론을 수행하지 않기 때문에 PyTorch가 필요 없습니다.


### ✅ 체크포인트 04

1. 이미지를 API에 전송할 때 Base64로 인코딩하는 이유는 무엇입니까?
   - **답변**: 이미지는 이진 데이터이므로, HTTP 요청의 JSON 본문에 직접 포함할 수 없습니다. Base64 인코딩을 통해 이미지를 텍스트 문자열로 변환하여 JSON 형태로 API에 안전하게 전송할 수 있습니다.
2. `response.raise_for_status()`는 어떤 역할을 합니까?
   - **답변**: HTTP 요청의 응답 상태 코드가 4xx (클라이언트 에러) 또는 5xx (서버 에러)일 경우 예외를 발생시킵니다. 이를 통해 네트워크 오류나 서버 오류를 쉽게 감지하고 처리할 수 있습니다.

Q3. 프론트엔드와 백엔드를 분리하는 핵심 이유 두 가지는?

[섹션 4: API 호출]
Q5. API 호출 실패 시 사용자에게 스택 트레이스가 아닌 메시지를 보여줘야 하는 이유는?

   - **답변**: 사용자에게 기술적인 스택 트레이스를 보여주는 것은 혼란을 줄 수 있고, 보안상 민감한 정보를 노출할 위험이 있습니다. 대신 친화적이고 이해하기 쉬운 에러 메시지를 제공하여 사용자가 문제를 파악하고 해결하는 데 도움을 줘야 합니다.


[섹션 5: 실습]
Q6. st.session_state에 결과를 저장하는 이유는?

   - **답변**: Streamlit은 사용자의 상호작용마다 스크립트 전체를 재실행하기 때문에, 재실행 시 위젯의 상태나 변수 값이 초기화됩니다. `st.session_state`에 결과를 저장하면 스크립트 재실행 후에도 이전 결과를 유지하고 UI에 표시할 수 있습니다.


Q7. 이미지를 API로 전달할 때 Base64 인코딩이 필요한 이유는?

   - **답변**: 이미지는 이진 데이터인데, HTTP 요청의 JSON 본문은 텍스트 형식이어야 합니다. Base64 인코딩은 이진 데이터를 아스키(ASCII) 문자열 형식으로 변환하여 JSON 본문에 안전하게 포함시켜 API로 전송할 수 있게 합니다.
